```python
def faktor(n):
  x=[]
  for i in range(1,n+1):
    if n%i==0:
      x.append(i)
  return x
```

```python
faktor(6)
```

```python
def prima(n):
  if len(faktor(n))==2:
    print(n,'bilangan prima')
  else:
    print(n,'bukan bilangan prima')
```

```python
prima(7)
```

```python
def fpb(a, b):
    for i in range(min(a, b), 0, -1):
        if a % i == 0 and b % i == 0:
            return i
```

```python
fpb(12,18)
```

```python
def fpb(a, b):
    faktor_sama = []
    for i in faktor(a):
        if i in faktor(b):
            faktor_sama.append(i)
    return max(faktor_sama)
```

```python
fpb(12,18)
```

```python
def relativeprima (a,b):
  if fpb(a,b)==1:
    print(a,'dan',b,'relative prima')
  else:
    print(a,'dan',b,'bukan relative prima')
```

```python
relativeprima(18,6)
```

```python
def kpk(a, b):
    for i in range(max(a, b), (a * b) + 1):
        if i % a == 0 and i % b == 0:
            return i
```

```python
kpk(2,5)
```
